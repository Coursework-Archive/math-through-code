## Question 16

**(a)**

Start with:
$$
y = e^x
$$

Reflect across the horizontal line $y = 4$.

Reflection rule across $y = c$:
$$
y_{\text{new}} = 2c - y
$$

Here $c = 4$, so:
$$
y_{\text{new}} = 2(4) - e^x
$$

$$
y_{\text{new}} = 8 - e^x
$$

$$\boxed{y = 8 - e^x}$$

**(b)**

Reflection Across the Line $x = a$

To reflect a point $(x,y)$ across the vertical line $x=a$, the midpoint in the $x$-direction must lie on $x=a$.

Thus,

$$
\frac{x + x_{\text{new}}}{2} = a
$$

Solving:

$$
x + x_{\text{new}} = 2a
$$

$$
x_{\text{new}} = 2a - x
$$

Reflection Across $x = 2$

Substitute $a=2$:

$$
x_{\text{new}} = 2(2) - x
$$

$$
x_{\text{new}} = 4 - x
$$

Therefore, reflecting $f(x)=a^x$ across $x=2$ gives:

$$
\boxed{\,y = e^{4 - x}\,}
$$

## Question 20

Let:
$$
f(x) = C b^x
$$

Given points:
$$
(-1, 3)
\quad \text{and} \quad
(1, \tfrac{4}{3})
$$

Substitute $x = -1$:
$$
3 = C b^{-1}
$$

Substitute $x = 1$:
$$
\frac{4}{3} = C b
$$

Divide the second equation by the first:
$$
\frac{\frac{4}{3}}{3} = \frac{Cb}{Cb^{-1}}
$$

$$
\frac{4}{9} = b^2
$$

$$
b = \frac{2}{3}
$$

Now solve for $C$:

Using $3 = C b^{-1}$ and $b = \frac{2}{3}$:

$$
3 = C \left(\frac{3}{2}\right)
$$

$$
C = 2
$$

Final function:
$$
f(x) = 2\left(\frac{2}{3}\right)^x
$$

$$
\boxed{f(x) = 2\left(\frac{2}{3}\right)^x}
$$

## Question 24

```python
x = sp.symbols("x")

display_latex(
    r"$\textbf{Graphs of the intersections } y = x^5 \textbf{ and } y = 5^x$"
)


pieces = [
    (x**5, (-2, 6, False, False)),
    (5**x, (-2, 6, False, False)),
]

plot_piecewise(
    pieces,
    var=x,
    xlim=(1.3, 1.85),
    ylim=(10, 20),
    labeled_points=[(1.76, 17.1, "(1.8, 17.1)")],
)
```

```python
plot_piecewise(
    pieces,
    var=x,
    xlim=(4.5, 5.5),
    ylim=(2500, 4000),
    labeled_points=[(5.0, 3125, "(5.0, 3125.0)")],
)
```
