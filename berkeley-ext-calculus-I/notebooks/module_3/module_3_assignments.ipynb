## Question 15

Find the limit or show that it does not exist.
$$
\lim_{x \to \infty} \frac{4x + 3}{5x - 1}
$$

Divide numerator and denominator by $x$:

$$
= \lim_{x \to \infty}
\frac{\frac{4x}{x} + \frac{3}{x}}
{\frac{5x}{x} - \frac{1}{x}}
$$

$$
= \lim_{x \to \infty}
\frac{4 + \frac{3}{x}}
{5 - \frac{1}{x}}
$$

As $x \to \infty$,
$$
\frac{3}{x} \to 0
\quad \text{and} \quad
\frac{1}{x} \to 0
$$

Therefore,

$$
\lim_{x \to \infty} \frac{4x + 3}{5x - 1}
=
\frac{4}{5}
$$

$$
\boxed{\frac{4}{5}}
$$

## Question 64

Given:
$$
y = x^2 (x^2 - 1)^2 (x + 2)
$$

Determine the leading term:

$$
x^2 (x^2 - 1)^2 (x + 2)
\sim x^2 \cdot x^4 \cdot x
= x^7
$$

The polynomial has:

- Degree: $7$ (odd)
- Leading coefficient: positive

Therefore,

$$
\boxed{\lim_{x \to \infty} y = \infty}
$$

$$
\boxed{\lim_{x \to -\infty} y = -\infty}
$$


Intercepts:

Solve each factor equal to zero.

$$
x^2 = 0 \Rightarrow x = 0 \quad (\text{multiplicity } 2)
$$

$$
x^2 - 1 = 0 \Rightarrow x = \pm 1 \quad (\text{multiplicity } 2)
$$

$$
x + 2 = 0 \Rightarrow x = -2 \quad (\text{multiplicity } 1)
$$

Thus the x-intercepts are:

$$
\boxed{(-2,0),\ (-1,0),\ (0,0),\ (1,0)}
$$

The y-intercept:

$$
y(0) = 0
$$

So,

$$
\boxed{(0,0)}
$$

Behavior at Each Intercept

- $x = -2$ (multiplicity 1) → **crosses**
- $x = -1$ (multiplicity 2) → **touches / bounces**
- $x = 0$ (multiplicity 2) → **touches / bounces**
- $x = 1$ (multiplicity 2) → **touches / bounces**


Sign Analysis

Since

$$
x^2 \ge 0, \qquad (x^2 - 1)^2 \ge 0
$$

the sign of $y$ is determined by $(x + 2)$.

If $x < -2$, then $x + 2 < 0$ so $y < 0$.

If $x > -2$, then $x + 2 > 0$ so $y > 0$ (except at intercepts).

Definition 1

The tangent line to the curve $y = f(x)$ at the point $P(a, f(a))$ is the line through $P$ with slope

$$
m = \lim_{x \to a} \frac{f(x) - f(a)}{x - a}

$$
f(-1 + h)

## Question 22

Given:
$$
f(x)=\frac{1}{\sqrt{2x+2}}, \qquad a=1
$$

Using Equation 5:
$$
f'(a)=\lim_{x\to a}\frac{f(x)-f(a)}{x-a}
=\lim_{x\to a}\frac{\frac{1}{\sqrt{2x+2}}-\frac{1}{\sqrt{2a+2}}}{x-a}
$$

Combine the fractions:
$$
f'(a)=\lim_{x\to a}\frac{\sqrt{2a+2}-\sqrt{2x+2}}{(x-a)\sqrt{2x+2}\sqrt{2a+2}}
$$

Multiply by the conjugate:
$$
f'(a)=\lim_{x\to a}
\frac{\sqrt{2a+2}-\sqrt{2x+2}}{(x-a)\sqrt{2x+2}\sqrt{2a+2}}
\cdot
\frac{\sqrt{2a+2}+\sqrt{2x+2}}{\sqrt{2a+2}+\sqrt{2x+2}}
$$

Simplify the numerator:
$$
(\sqrt{2a+2})^2-(\sqrt{2x+2})^2=(2a+2)-(2x+2)=2(a-x)=-2(x-a)
$$

So,
$$
f'(a)=\lim_{x\to a}
\frac{-2(x-a)}{(x-a)\sqrt{2x+2}\sqrt{2a+2}\left(\sqrt{2a+2}+\sqrt{2x+2}\right)}
$$

Cancel $(x-a)$:
$$
f'(a)=\lim_{x\to a}
\frac{-2}{\sqrt{2x+2}\sqrt{2a+2}\left(\sqrt{2a+2}+\sqrt{2x+2}\right)}
$$

Let $x\to a$:
$$
f'(a)=\frac{-2}{(\sqrt{2a+2})(\sqrt{2a+2})(2\sqrt{2a+2})}
=\frac{-2}{(2a+2)(2\sqrt{2a+2})}
=-\frac{1}{(2a+2)\sqrt{2a+2}}
$$

Evaluate at $a=1$:
$$
f'(1)=-\frac{1}{(2\cdot 1+2)\sqrt{2\cdot 1+2}}
=-\frac{1}{4\cdot 2}
=-\frac{1}{8}
$$

$$
\boxed{f'(1)=-\frac{1}{8}}
$$

## Question 25

Given:
$$
f(t)=\frac{1}{t^2+1}
$$

Using Equation 5:
$$
f'(a)=\lim_{t\to a}\frac{f(t)-f(a)}{t-a}
=\lim_{t\to a}\frac{\frac{1}{t^2+1}-\frac{1}{a^2+1}}{t-a}
$$

Combine the fractions:
$$
f'(a)=\lim_{t\to a}
\frac{(a^2+1)-(t^2+1)}{(t-a)(t^2+1)(a^2+1)}
$$

Simplify the numerator:
$$
(a^2+1)-(t^2+1)=a^2-t^2
$$

Factor difference of squares:
$$
a^2-t^2=(a-t)(a+t)
$$

So,
$$
f'(a)=\lim_{t\to a}
\frac{(a-t)(a+t)}{(t-a)(t^2+1)(a^2+1)}
$$

Substitute:
$$
f'(a)=\lim_{t\to a}
\frac{-(t-a)(a+t)}{(t-a)(t^2+1)(a^2+1)}
$$

Cancel $(t-a)$:
$$
f'(a)=\lim_{t\to a}
\frac{-(a+t)}{(t^2+1)(a^2+1)}
$$

Now let $t\to a$:
$$
f'(a)=\frac{-(a+a)}{(a^2+1)(a^2+1)}
=\frac{-2a}{(a^2+1)^2}
$$

$$
\boxed{f'(a)=-\frac{2a}{(a^2+1)^2}}
$$

## Question 9


```python
x = sp.symbols("x")
display_latex(r"f")
f_left  = (x + sp.Rational(1,2))**2 - 1
a = 1
f_a = sp.simplify(f_left.subs(x, a))         # value at the join
m = -1                                       # slope of the line
f_right = f_a + m*(x - a)                    # line through (a, f(a))

f = sp.Piecewise((f_left, x <= a), (f_right, True))

pieces = [(f, (-3, 3, False, False))]

plot_piecewise(
    pieces,
    xlim=(-3, 3),
    ylim=(-3, 4),
    style_origin=True,
    labeled_points=[
        (-2, float(f.subs(x, -2)), "f'(-2)=-25/16"),
        (-1.5, float(f.subs(x, -1.5)), "f'(-3/2)=-5/4"),
        (-0.5, float(f.subs(x, -0.5)), "f'(-1/2)=0"),
        (0.5, float(f.subs(x, 0.5)), "f'(1/2)=1"),
        (1, float(f.subs(x, 1)), "f'(1)=DNE"),
        (2, float(f.subs(x, 2)), "f'(2)=-1"),
    ],
    title=None
)


display_latex(r"f'")
a = 1
fp_left = sp.diff(f_left, x)   # derivative of left piece
fp_right = sp.diff(f_right, x) # derivative of right piece (a constant)

pieces = [
    (fp_left,  (-3, a, True,  False)),  # ... up to a, open at a
    (fp_right, (a,  3, False, False)),   # from a onward, open at a
]

open_points = [
    (a, float(fp_left.subs(x, a))),     # left-hand derivative value
    (a, float(fp_right.subs(x, a))),    # right-hand derivative value
]


plot_piecewise(
    pieces,
    xlim=(-3, 3),
    ylim=(-3, 4),
    title=None,
    style_origin=True,
    open_points=open_points
)

```

\frac{mx + mh + b - (mx + b)}{h}

\frac{mh}{h}
=
m
$$

$$
\boxed{f'(x) = m}
