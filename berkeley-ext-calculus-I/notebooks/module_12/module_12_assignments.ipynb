## Problem 13

Find the volume obtained by rotating the region bounded by

$$
y=\sqrt{x-1},\qquad y=0,\qquad x=5
$$

about the x-axis.

Find the left endpoint

Set

$$
\sqrt{x-1}=0
$$

Then

$$
x-1=0
$$

so

$$
x=1
$$

Therefore,

$$
1\le x\le 5
$$

Radius of a disk

Since the region is rotated about the x-axis,

$$
r(x)=y=\sqrt{x-1}
$$

Cross-sectional area:

$$
A(x)=\pi r^2
$$

Thus

$$
A(x)=\pi(\sqrt{x-1})^2
$$

and

$$
A(x)=\pi(x-1)
$$

Volume integral

$$
V=\int_1^5 A(x)\,dx
$$

$$
V=\pi\int_1^5(x-1)\,dx
$$

Integrate:

$$
V=\pi\left[\frac{x^2}{2}-x\right]_1^5
$$

Evaluate:

At $x=5$,

$$
\frac{25}{2}-5=\frac{15}{2}
$$

At $x=1$,

$$
\frac12-1=-\frac12
$$

Therefore,

$$
V=\pi\left(\frac{15}{2}-\left(-\frac12\right)\right)
$$

$$
V=\pi\left(\frac{16}{2}\right)
$$

$$
\boxed{V=8\pi}
$$



```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

# Problem 13:
# y = sqrt(x - 1), y = 0, x = 5
# Rotate about the x-axis

x = np.linspace(1, 5, 400)
y = np.sqrt(x - 1)

# Typical disk location
x_disk = 3.5
r_disk = np.sqrt(x_disk - 1)

# -----------------------------
# Graph 1: Region
# -----------------------------
plt.figure(figsize=(8, 4))

plt.plot(x, y, label=r"$y=\sqrt{x-1}$")
plt.plot(x, np.zeros_like(x), label=r"$y=0$")
plt.axvline(5, linestyle="--", label=r"$x=5$")

plt.fill_between(x, 0, y, alpha=0.3, label="Region")

plt.scatter([1, 5], [0, np.sqrt(4)])
plt.text(1, -0.15, r"$(1,0)$")
plt.text(5, 2.1, r"$x=5$")

plt.title(r"Region bounded by $y=\sqrt{x-1}$, $y=0$, and $x=5$")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0)
plt.axvline(0)
plt.legend()
plt.grid(True)
plt.show()


# -----------------------------
# Graph 2: Solid of revolution with typical disk
# -----------------------------
plt.figure(figsize=(9, 4))

# Top and bottom outlines of solid
plt.plot(x, y, label=r"$y=\sqrt{x-1}$")
plt.plot(x, -y, label=r"$y=-\sqrt{x-1}$")

# Fill the solid silhouette
plt.fill_between(x, -y, y, alpha=0.2)

# Draw a typical disk as an ellipse
disk = Ellipse(
    (x_disk, 0),
    width=0.25,
    height=2 * r_disk,
    fill=False,
    linewidth=2,
    label="Typical disk"
)
plt.gca().add_patch(disk)

# Radius line
plt.plot([x_disk, x_disk], [0, r_disk], linestyle="--")
plt.text(x_disk + 0.1, r_disk / 2, r"$r=\sqrt{x-1}$")

# Right end disk
end_disk = Ellipse(
    (5, 0),
    width=0.35,
    height=2 * np.sqrt(5 - 1),
    fill=False,
    linewidth=2
)
plt.gca().add_patch(end_disk)

plt.title(r"Solid formed by rotating about the x-axis")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0)
plt.axvline(0)
plt.legend()
plt.grid(True)
plt.axis("equal")
plt.show()
```

$$

$$
V=\pi\left[\frac12e^{2x}\right]_{-1}^{1}

$$

## Problem 15

$$
x=2\sqrt y,\qquad x=0,\qquad y=9
$$

about the y-axis.

Bounds

Since the curve begins at the origin,

$$
0\le y\le 9

Radius

$$

$$

## Problem 17

Find the volume obtained by rotating

$$
y=x^2,\qquad y=2x
$$

about the y-axis.

Find the points of intersection

Set

$$
x^2=2x
$$

$$
x(x-2)=0
$$

Hence,

$$
x=0,\qquad x=2
$$

Solve each equation for $x$:

From

$$
y=x^2
$$

we obtain

$$
x=\sqrt y
$$

From

$$
y=2x
$$

we obtain

$$
x=\frac y2
$$

Since the axis of rotation is the y-axis,

- outer radius:

$$
R(y)=\sqrt y
$$

- inner radius:

$$
r(y)=\frac y2
$$

Bounds

The corresponding y-values are

$$
y=0,\qquad y=4
$$

Washer area

$$
A(y)=\pi(R^2-r^2)
$$

Therefore,

$$
A(y)=\pi\left[(\sqrt y)^2-\left(\frac y2\right)^2\right]
$$

$$
A(y)=\pi\left(y-\frac{y^2}{4}\right)
$$

Volume integral

$$
V=\pi\int_0^4\left(y-\frac{y^2}{4}\right)dy
$$

Integrate:

$$
V=\pi\left[\frac{y^2}{2}-\frac{y^3}{12}\right]_0^4
$$

Evaluate:

$$
V=\pi\left(8-\frac{64}{12}\right)
$$

$$
V=\pi\left(8-\frac{16}{3}\right)
$$


$$
\boxed{V=\frac{8\pi}{3}}
$$



```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

# Problem 17:
# y = x^2, y = 2x
# Rotate about the y-axis

y = np.linspace(0, 4, 400)

# Solve both curves for x
x_outer = np.sqrt(y)   # from y = x^2
x_inner = y / 2        # from y = 2x

# Typical washer location
y_washer = 2.5
R = np.sqrt(y_washer)
r = y_washer / 2

# -----------------------------
# Graph 1: Region
# -----------------------------
plt.figure(figsize=(6, 6))

plt.plot(x_outer, y, label=r"$y=x^2$")
plt.plot(x_inner, y, label=r"$y=2x$")

plt.fill_betweenx(y, x_inner, x_outer, alpha=0.3, label="Region")

plt.title(r"Region bounded by $y=x^2$ and $y=2x$")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0)
plt.axvline(0, label="y-axis")
plt.legend()
plt.grid(True)
plt.show()


# -----------------------------
# Graph 2: Solid of revolution with typical washer
# -----------------------------
plt.figure(figsize=(7, 7))

# Outer boundary after rotation
plt.plot(x_outer, y, label=r"$x=\sqrt{y}$")
plt.plot(-x_outer, y, label=r"$x=-\sqrt{y}$")

# Inner boundary after rotation
plt.plot(x_inner, y, linestyle="--", label=r"$x=\frac{y}{2}$")
plt.plot(-x_inner, y, linestyle="--", label=r"$x=-\frac{y}{2}$")

# Fill outer solid silhouette
plt.fill_betweenx(y, -x_outer, x_outer, alpha=0.15)

# Typical washer outer edge
outer_washer = Ellipse(
    (0, y_washer),
    width=2 * R,
    height=0.28,
    fill=False,
    linewidth=2,
    label="Outer radius"
)
plt.gca().add_patch(outer_washer)

# Typical washer inner edge
inner_washer = Ellipse(
    (0, y_washer),
    width=2 * r,
    height=0.14,
    fill=False,
    linewidth=2,
    label="Inner radius"
)
plt.gca().add_patch(inner_washer)

# Radius lines
plt.plot([0, R], [y_washer, y_washer], linestyle="--")
plt.plot([0, r], [y_washer - 0.08, y_washer - 0.08], linestyle="--")

plt.text(R + 0.05, y_washer, r"$R=\sqrt{y}$")
plt.text(r + 0.05, y_washer - 0.25, r"$r=\frac{y}{2}$")

plt.title(r"Solid formed by rotating about the y-axis")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0)
plt.axvline(0, label="axis of rotation")
plt.legend()
plt.grid(True)
plt.axis("equal")
plt.show()
```

## Problem 19

Find the volume obtained by rotating

$$
y=x^3,\qquad y=\sqrt x
$$

about the x-axis.

Find the points of intersection

Set

$$
x^3=\sqrt x
$$

Square both sides:

$$
x^6=x
$$

$$
x(x^5-1)=0
$$

Hence,

$$
x=0,\qquad x=1
$$

Radii

Outer radius:

$$
R(x)=\sqrt x
$$

Inner radius:

$$
r(x)=x^3
$$

Washer area

$$
A(x)=\pi(R^2-r^2)
$$

Thus,

$$
A(x)=\pi\left[(\sqrt x)^2-(x^3)^2\right]
$$

$$
A(x)=\pi(x-x^6)
$$

Volume integral

$$
V=\pi\int_0^1(x-x^6)\,dx
$$

Integrate:

$$
V=\pi\left[\frac{x^2}{2}-\frac{x^7}{7}\right]_0^1
$$

Evaluate:

$$
V=\pi\left(\frac12-\frac17\right)
$$

$$
V=\pi\left(\frac{7-2}{14}\right)
$$

$$
\boxed{V=\frac{5\pi}{14}}
$$



```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

# Problem 19:
# y = x^3, y = sqrt(x)
# Rotate about the x-axis

x = np.linspace(0, 1, 400)

y_outer = np.sqrt(x)
y_inner = x**3

# Typical washer location
x_washer = 0.55
R = np.sqrt(x_washer)
r = x_washer**3

# -----------------------------
# Graph 1: Region
# -----------------------------
plt.figure(figsize=(7, 5))

plt.plot(x, y_outer, label=r"$y=\sqrt{x}$")
plt.plot(x, y_inner, label=r"$y=x^3$")

plt.fill_between(x, y_inner, y_outer, alpha=0.3, label="Region")

plt.title(r"Region bounded by $y=x^3$ and $y=\sqrt{x}$")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0)
plt.axvline(0)
plt.legend()
plt.grid(True)
plt.show()


# -----------------------------
# Graph 2: Solid of revolution with typical washer
# -----------------------------
plt.figure(figsize=(8, 5))

# Outer boundary after rotation
plt.plot(x, y_outer, label=r"$y=\sqrt{x}$")
plt.plot(x, -y_outer, label=r"$y=-\sqrt{x}$")

# Inner boundary after rotation
plt.plot(x, y_inner, linestyle="--", label=r"$y=x^3$")
plt.plot(x, -y_inner, linestyle="--", label=r"$y=-x^3$")

# Fill outer silhouette
plt.fill_between(x, -y_outer, y_outer, alpha=0.15)

# Typical washer outer edge
outer_washer = Ellipse(
    (x_washer, 0),
    width=0.08,
    height=2 * R,
    fill=False,
    linewidth=2,
    label="Outer radius"
)
plt.gca().add_patch(outer_washer)

# Typical washer inner edge
inner_washer = Ellipse(
    (x_washer, 0),
    width=0.04,
    height=2 * r,
    fill=False,
    linewidth=2,
    label="Inner radius"
)
plt.gca().add_patch(inner_washer)

# Radius lines
plt.plot([x_washer, x_washer], [0, R], linestyle="--")
plt.plot([x_washer + 0.015, x_washer + 0.015], [0, r], linestyle="--")

plt.text(x_washer + 0.04, R / 2, r"$R=\sqrt{x}$")
plt.text(x_washer + 0.04, r / 2, r"$r=x^3$")

plt.title(r"Solid formed by rotating about the x-axis")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, label="axis of rotation")
plt.axvline(0)
plt.legend()
plt.grid(True)
plt.axis("equal")
plt.show()
```

## Problem 20

Find the volume obtained by rotating

$$
x=2-y^2,\qquad x=y^4
$$

about the y-axis.

Find the points of intersection

Set

$$
2-y^2=y^4
$$

Rearrange:

$$
y^4+y^2-2=0
$$

Let

$$
u=y^2
$$

Then

$$
u^2+u-2=0
$$

Factor:

$$
(u+2)(u-1)=0
$$

Since

$$
u=y^2\ge0,
$$

we obtain

$$
y^2=1
$$

Thus,

$$
y=\pm1
$$

Radii

Outer radius:

$$
R(y)=2-y^2
$$

Inner radius:

$$
r(y)=y^4
$$

Washer area

$$
A(y)=\pi(R^2-r^2)
$$

Therefore,

$$
A(y)=\pi\left[(2-y^2)^2-(y^4)^2\right]
$$

Expand:

$$
(2-y^2)^2=4-4y^2+y^4
$$

Hence,

$$
A(y)=\pi\left(4-4y^2+y^4-y^8\right)
$$

Volume integral

$$
V=\pi\int_{-1}^{1}(4-4y^2+y^4-y^8)\,dy
$$

Integrate:

$$
V=\pi\left[4y-\frac{4}{3}y^3+\frac15y^5-\frac19y^9\right]_{-1}^{1}
$$

Evaluate:

At $y=1$,

$$
4-\frac43+\frac15-\frac19=\frac{124}{45}
$$

At $y=-1$,

$$
-4+\frac43-\frac15+\frac19=-\frac{124}{45}
$$

Therefore,

$$
V=\pi\left(\frac{124}{45}-\left(-\frac{124}{45}\right)\right)
$$

$$
\boxed{V=\frac{248\pi}{45}}
$$

## Problem 24

Find the volume obtained by rotating

$$
y=\sin x,\qquad y=\cos x,\qquad 0\le x\le\frac{\pi}{4}
$$

about

$$
y=-1.
$$

Since

$$
\cos x\ge\sin x
$$

on

$$
0\le x\le\frac{\pi}{4},
$$

the radii are

$$
R(x)=1+\cos x
$$

and

$$
r(x)=1+\sin x.
$$

Washer area

$$
A(x)=\pi(R^2-r^2)
$$

Thus,

$$
A(x)=\pi\left[(1+\cos x)^2-(1+\sin x)^2\right]
$$

Simplify:

$$
A(x)=\pi\left(2\cos x-2\sin x+\cos^2x-\sin^2x\right)
$$

Using

$$
\cos^2x-\sin^2x=\cos(2x),
$$

we have

$$
A(x)=\pi\left(2\cos x-2\sin x+\cos(2x)\right)
$$

Volume integral

$$
V=\pi\int_0^{\pi/4}
\left(2\cos x-2\sin x+\cos(2x)\right)dx
$$

Integrate:

$$
V=\pi\left[
2\sin x+2\cos x+\frac12\sin(2x)
\right]_0^{\pi/4}
$$

Evaluate:

$$
V=\pi\left[\left(2\sqrt2+\frac12\right)-2\right]
$$

$$
V=\pi\left(2\sqrt2-\frac32\right)
$$

Therefore,

$$
\boxed{
V=\frac{\pi}{2}(4\sqrt2-3)
}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

# Problem 24:
# y = sin(x), y = cos(x), 0 <= x <= pi/4
# Rotate about y = -1

x = np.linspace(0, np.pi / 4, 400)

y_top = np.cos(x)
y_bottom = np.sin(x)

axis = -1

# Radii measured from y = -1
R = y_top - axis      # 1 + cos(x)
r = y_bottom - axis   # 1 + sin(x)

# Typical washer location
x_washer = np.pi / 6
R_washer = np.cos(x_washer) - axis
r_washer = np.sin(x_washer) - axis

# -----------------------------
# Graph 1: Region
# -----------------------------
plt.figure(figsize=(8, 5))

plt.plot(x, y_top, label=r"$y=\cos x$")
plt.plot(x, y_bottom, label=r"$y=\sin x$")
plt.axhline(axis, linestyle="--", label=r"axis of rotation: $y=-1$")
plt.axvline(0, linestyle="--", label=r"$x=0$")
plt.axvline(np.pi / 4, linestyle="--", label=r"$x=\frac{\pi}{4}$")

plt.fill_between(x, y_bottom, y_top, alpha=0.3, label="Region")

plt.title(r"Region bounded by $y=\sin x$ and $y=\cos x$")
plt.xlabel("x")
plt.ylabel("y")
plt.xticks(
    [0, np.pi / 8, np.pi / 4],
    [r"$0$", r"$\frac{\pi}{8}$", r"$\frac{\pi}{4}$"]
)
plt.axhline(0)
plt.axvline(0)
plt.legend()
plt.grid(True)
plt.show()


# -----------------------------
# Graph 2: Solid of revolution with typical washer
# -----------------------------
plt.figure(figsize=(9, 6))

# Upper and lower outlines after rotating about y = -1
outer_top = axis + R
outer_bottom = axis - R

inner_top = axis + r
inner_bottom = axis - r

plt.plot(x, outer_top, label=r"outer: $y=\cos x$")
plt.plot(x, outer_bottom, label=r"rotated outer boundary")
plt.plot(x, inner_top, linestyle="--", label=r"inner: $y=\sin x$")
plt.plot(x, inner_bottom, linestyle="--", label=r"rotated inner boundary")

# Fill the outer silhouette
plt.fill_between(x, outer_bottom, outer_top, alpha=0.15)

# Axis of rotation
plt.axhline(axis, linestyle="--", label=r"axis of rotation: $y=-1$")

# Typical washer outer edge
outer_washer = Ellipse(
    (x_washer, axis),
    width=0.08,
    height=2 * R_washer,
    fill=False,
    linewidth=2,
    label="Outer radius"
)
plt.gca().add_patch(outer_washer)

# Typical washer inner edge
inner_washer = Ellipse(
    (x_washer, axis),
    width=0.04,
    height=2 * r_washer,
    fill=False,
    linewidth=2,
    label="Inner radius"
)
plt.gca().add_patch(inner_washer)

# Radius lines measured from y = -1
plt.plot([x_washer, x_washer], [axis, axis + R_washer], linestyle="--")
plt.plot([x_washer + 0.015, x_washer + 0.015], [axis, axis + r_washer], linestyle="--")

plt.text(x_washer + 0.04, axis + R_washer / 2, r"$R=1+\cos x$")
plt.text(x_washer + 0.04, axis + r_washer / 2, r"$r=1+\sin x$")

plt.title(r"Solid formed by rotating about $y=-1$")
plt.xlabel("x")
plt.ylabel("y")
plt.xticks(
    [0, np.pi / 8, np.pi / 4],
    [r"$0$", r"$\frac{\pi}{8}$", r"$\frac{\pi}{4}$"]
)
plt.axhline(0)
plt.axvline(0)
plt.legend()
plt.grid(True)
plt.axis("equal")
plt.show()
```

about the y-axis using cylindrical shells.

Therefore,

$$
V=2\pi\int_0^4x\sqrt{x}\,dx
$$

Rewrite:

$$
x\sqrt{x}=x^{3/2}

Hence,

$$
V=2\pi\int_0^4x^{3/2}\,dx

V=2\pi\left[\frac{2}{5}x^{5/2}\right]_0^4
$$

Since

$$
4^{5/2}=32,
$$

we obtain

$$
V=2\pi\left(\frac{2}{5}(32)\right)
$$

Therefore,

$$
\boxed{V=\frac{128\pi}{5}}
