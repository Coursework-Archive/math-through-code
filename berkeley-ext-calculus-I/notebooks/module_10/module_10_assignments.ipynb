## Question 4

Given:

$$
f(x)=\frac{1}{x}, \quad 1 \le x \le 2
$$

There are four subintervals, so:

$$
\Delta x=\frac{2-1}{4}=\frac{1}{4}
$$

**(a)** Right Endpoints

The subintervals are:

$$
[1,\frac{5}{4}], \left[\frac{5}{4},\frac{3}{2}\right], \left[\frac{3}{2},\frac{7}{4}\right], \left[\frac{7}{4},2\right]
$$

The right endpoints are:

$$
\frac{5}{4}, \frac{3}{2}, \frac{7}{4}, 2
$$

The Riemann sum is:

$$
\frac{1}{4}\left[f\left(\frac{5}{4}\right)+f\left(\frac{3}{2}\right)+f\left(\frac{7}{4}\right)+f(2)\right]
$$

$$
=\frac{1}{4}\left[\frac{4}{5}+\frac{2}{3}+\frac{4}{7}+\frac{1}{2}\right]
$$

$$
=0.634524
$$

$$
\boxed{0.634524}
$$

This Riemann sum represents an estimate of the area under the curve

$$
y=\frac{1}{x}
$$

from \(x=1\) to \(x=2\), using four rectangles whose heights are determined by right endpoints.



```python
import numpy as np
import matplotlib.pyplot as plt

# Function
def f(x):
    return 1 / x

# Interval [1, 2]
a = 1
b = 2

# Number of rectangles
n = 4

# Width of each subinterval
dx = (b - a) / n

# x-values for smooth curve
x = np.linspace(a, b, 500)
y = f(x)

# Right endpoints
x_right = np.array([a + i * dx for i in range(1, n + 1)])

# Left edges of rectangles
x_left = np.array([a + i * dx for i in range(n)])

# Heights using right endpoints
heights = f(x_right)

# Create figure
plt.figure(figsize=(8, 5))

# Plot function
plt.plot(x, y, label=r"$f(x)=\frac{1}{x}$")

# Draw rectangles
plt.bar(
    x_left,
    heights,
    width=dx,
    align="edge",
    edgecolor="black",
    alpha=0.4,
    label="Right-endpoint rectangles"
)

# Plot right endpoint points
plt.scatter(
    x_right,
    heights,
    zorder=3,
    label="Right endpoints"
)

# Labels and title
plt.title(r"Riemann Sum for $f(x)=\frac{1}{x}$ on $[1,2]$")
plt.xlabel("x")
plt.ylabel("y")

# Nice x-axis labels
plt.xticks(
    [1, 1.25, 1.5, 1.75, 2],
    [r"$1$", r"$\frac{5}{4}$", r"$\frac{3}{2}$", r"$\frac{7}{4}$", r"$2$"]
)

plt.grid(True)
plt.legend()

plt.show()
```


**(b)** Midpoints

The midpoints are:

$$
\frac{9}{8}, \frac{11}{8}, \frac{13}{8}, \frac{15}{8}
$$

The Riemann sum is:

$$
\frac{1}{4}\left[f\left(\frac{9}{8}\right)+f\left(\frac{11}{8}\right)+f\left(\frac{13}{8}\right)+f\left(\frac{15}{8}\right)\right]
$$

$$
=\frac{1}{4}\left[\frac{8}{9}+\frac{8}{11}+\frac{8}{13}+\frac{8}{15}\right]
$$

$$
=0.691220
$$

$$
\boxed{0.691220}
$$

This Riemann sum represents an estimate of the area under the curve

$$
y=\frac{1}{x}
$$

from \(x=1\) to \(x=2\), using four rectangles whose heights are determined by the midpoint of each subinterval.

```python
import numpy as np
import matplotlib.pyplot as plt

# Function
def f(x):
    return 1 / x

# Interval [1, 2]
a = 1
b = 2

# Number of rectangles
n = 4

# Width of each subinterval
dx = (b - a) / n

# Smooth curve
x = np.linspace(a, b, 500)
y = f(x)

# Left edges of rectangles
x_left = np.array([a + i * dx for i in range(n)])

# Midpoints
x_mid = x_left + dx / 2

# Heights at midpoints
heights = f(x_mid)

# Create figure
plt.figure(figsize=(8, 5))

# Plot function
plt.plot(x, y, linewidth=2, label=r"$f(x)=\frac{1}{x}$")

# Draw rectangles
plt.bar(
    x_left,
    heights,
    width=dx,
    align="edge",
    edgecolor="black",
    alpha=0.35,
    label="Midpoint rectangles"
)

# Add dotted midpoint lines
plt.vlines(
    x_mid,
    ymin=0,
    ymax=heights,
    linestyles="dashed",
    linewidth=1.5
)

# Plot midpoint dots
plt.scatter(
    x_mid,
    heights,
    zorder=3,
    label="Midpoints"
)

# Labels and title
plt.title(r"Midpoint Riemann Sum for $f(x)=\frac{1}{x}$ on $[1,2]$")
plt.xlabel("x")
plt.ylabel("y")

# Better x-axis labels
plt.xticks(
    [1, 1.125, 1.25, 1.375, 1.5, 1.625, 1.75, 1.875, 2],
    [
        r"$1$",
        r"$\frac{9}{8}$",
        r"$\frac{5}{4}$",
        r"$\frac{11}{8}$",
        r"$\frac{3}{2}$",
        r"$\frac{13}{8}$",
        r"$\frac{7}{4}$",
        r"$\frac{15}{8}$",
        r"$2$"
    ]
)

plt.grid(True, alpha=0.3)
plt.legend()

plt.show()
```

## Question 18

For

$$
\int_0^2 e^{-x^2}\,dx
$$

the left and right Riemann sums are:

| $n$ | $L_n$ | $R_n$ |
|---:|---:|---:|
| 5 | 1.0775 | 0.6848 |
| 10 | 0.9800 | 0.7837 |
| 50 | 0.9017 | 0.8624 |
| 100 | 0.8919 | 0.8723 |

Since

$$
e^{-x^2}
$$

is decreasing on

$$
[0,2]
$$

the integral must lie between the left and right sums:

$$
R_n
<
\int_0^2 e^{-x^2}\,dx
<
L_n
$$

Using \(n=100\):

$$
0.8723 < \int_0^2 e^{-x^2}\,dx < 0.8919
$$

So the value of the integral appears to approach:

$$
\boxed{0.8821}
$$

For

$$
\int_{-1}^{2} e^{-x^2}\,dx
$$

a similar convergence statement can be made: as \(n\) gets larger, both the left and right Riemann sums approach the value of the definite integral.

However,

$$
e^{-x^2}
$$

is not decreasing on the entire interval

$$
[-1,2]
$$

so we cannot simply say that the right sum is always an underestimate and the left sum is always an overestimate.
