## Exercise 2

Find:

$$
\int \tan^2(x)\,dx
$$

Use the identity:

$$
\tan^2(x)=\sec^2(x)-1
$$

Then

$$
\int \tan^2(x)\,dx
=
\int(\sec^2(x)-1)\,dx
$$

$$
=
\int \sec^2(x)\,dx-\int 1\,dx
$$

Using

$$
\int \sec^2(x)\,dx=\tan(x)
$$

and

$$
\int 1\,dx=x
$$

gives

$$
\boxed{\tan(x)-x+C}
$$

## Exercise 57

Evaluate

$$
\int e^{\cos x}\sin x\,dx
$$

Let

$$
u=\cos x
$$

Then

$$
du=-\sin x\,dx
$$

or

$$
-\;du=\sin x\,dx
$$

Substitute:

$$
-\int e^u\,du
$$

Integrate:

$$
-e^u+C
$$

Substitute back:

$$
\boxed{-e^{\cos x}+C}
$$


```python
import numpy as np
import matplotlib.pyplot as plt

# Domain
x = np.linspace(-2*np.pi, 2*np.pi, 2000)

# Function
f = np.exp(np.cos(x)) * np.sin(x)

# Antiderivative (C = 0)
F = -np.exp(np.cos(x))

# Create figure
plt.figure(figsize=(10, 6))

# Plot curves
plt.plot(x, f, linewidth=2,
         label=r'$f(x)=e^{\cos x}\sin x$')

plt.plot(x, F, linewidth=2,
         label=r'$F(x)=-e^{\cos x}$')

# π tick locations
pi_ticks = [
    -2*np.pi,
    -3*np.pi/2,
    -np.pi,
    -np.pi/2,
    0,
    np.pi/2,
    np.pi,
    3*np.pi/2,
    2*np.pi
]

# π tick labels
pi_labels = [
    r'$-2\pi$',
    r'$-\frac{3\pi}{2}$',
    r'$-\pi$',
    r'$-\frac{\pi}{2}$',
    r'$0$',
    r'$\frac{\pi}{2}$',
    r'$\pi$',
    r'$\frac{3\pi}{2}$',
    r'$2\pi$'
]

plt.xticks(pi_ticks, pi_labels)

# Axes and grid
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

plt.xlabel(r'$x$')
plt.ylabel(r'$y$')
plt.title(
    r'Function $e^{\cos x}\sin x$ and Antiderivative $-e^{\cos x}$'
)

plt.grid(True)
plt.legend()

plt.show()
```

## Problem 14

Given

$$
y=\cos x,
\qquad
y=e^x,
\qquad
x=\frac{\pi}{2},
$$

the curves intersect at

$$
x=0,
\qquad
(0,1).
$$

Since

$$
e^x>\cos x
\qquad
\text{for }
0\le x\le\frac{\pi}{2},
$$

the area is

$$
A=
\int_0^{\pi/2}
(e^x-\cos x)\,dx
$$

Integrating,

$$
A=
\left[
e^x-\sin x
\right]_0^{\pi/2}
$$

$$
=
\left(e^{\pi/2}-1\right)-(1-0)
$$

Therefore,

$$
\boxed{
A=e^{\pi/2}-2
}
$$



```python
import numpy as np
import matplotlib.pyplot as plt

# Problem 14:
# y = cos(x), y = e^x, x = pi/2
# Region is from x = 0 to x = pi/2

x = np.linspace(0, np.pi / 2, 400)

y_top = np.exp(x)
y_bottom = np.cos(x)

plt.figure(figsize=(8, 5))

# Plot curves
plt.plot(x, y_top, label=r"$y=e^x$")
plt.plot(x, y_bottom, label=r"$y=\cos x$")

# Shade enclosed region
plt.fill_between(
    x,
    y_bottom,
    y_top,
    alpha=0.3,
    label="Enclosed region"
)

# Draw boundary lines
plt.axvline(0, linestyle="--")
plt.axvline(np.pi / 2, linestyle="--")

# Mark intersection point
plt.scatter([0], [1], color='black')
plt.text(0.03, 1.05, r"$(0,1)$")

# Label right boundary
plt.text(np.pi / 2 + 0.03, 0.2, r"$x=\frac{\pi}{2}$", rotation=90)

# Add x- and y-axes
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# Set x-axis ticks in terms of pi
xticks = [0, np.pi/4, np.pi/2]
xtick_labels = [
    r"$0$",
    r"$\frac{\pi}{4}$",
    r"$\frac{\pi}{2}$"
]
plt.xticks(xticks, xtick_labels)

plt.title(r"Region Enclosed by $y=e^x$, $y=\cos x$, and $x=\frac{\pi}{2}$")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")

plt.xlim(-0.2, np.pi / 2 + 0.3)
plt.ylim(-0.2, np.exp(np.pi / 2) + 0.5)

plt.grid(True)
plt.legend()

plt.show()
```

## Problem 16

Given

$$
y=x^2-4x,
\qquad
y=2x,
$$

find the points of intersection:

$$
x^2-4x=2x
$$

$$
x^2-6x=0
$$

$$
x(x-6)=0
$$

Thus,

$$
x=0,\qquad x=6.
$$

The top curve is

$$
y=2x
$$

and the bottom curve is

$$
y=x^2-4x.
$$

Therefore,

$$
A=
\int_0^6
\left(2x-(x^2-4x)\right)\,dx
$$

$$
=
\int_0^6
(6x-x^2)\,dx
$$

Integrating,

$$
A=
\left[
3x^2-\frac{x^3}{3}
\right]_0^6
$$

Evaluating,

$$
A=
108-72
=
36
$$

$$
\boxed{A=36}
$$



```python
import numpy as np
import matplotlib.pyplot as plt

# Problem 16:
# y = x^2 - 4x, y = 2x
# Region is from x = 0 to x = 6

x = np.linspace(0, 6, 400)

y_top = 2 * x
y_bottom = x**2 - 4 * x

plt.figure(figsize=(8, 5))

# Plot curves
plt.plot(x, y_top, label=r"$y=2x$")
plt.plot(x, y_bottom, label=r"$y=x^2-4x$")

# Shade enclosed region
plt.fill_between(
    x,
    y_bottom,
    y_top,
    alpha=0.3,
    label="Enclosed region"
)

# Mark intersection points
plt.scatter([0, 6], [0, 12], color='black')
plt.text(0.15, 0.2, r"$(0,0)$")
plt.text(5.2, 12.5, r"$(6,12)$")

# Draw coordinate axes
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

plt.title(r"Region Enclosed by $y=2x$ and $y=x^2-4x$")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")

plt.xlim(-0.5, 6.5)
plt.ylim(-5, 14)

# Integer tick marks
plt.xticks(np.arange(0, 7, 1))
plt.yticks(np.arange(-4, 15, 2))

plt.grid(True)
plt.legend()

plt.show()
```

## Problem 20

$$

$$

$$
x(x-2)=0
$$

Therefore,

$$
x=0,
\qquad
x=2.
$$

For

$$
0\le x\le2,
$$

the upper curve is

$$
y=4x-x^2
$$

and the lower curve is

$$
y=x^2.
$$

Hence,

$$
A=
\int_0^2
\left[(4x-x^2)-x^2\right]dx
$$

$$
=
\int_0^2
(4x-2x^2)\,dx
$$

Integrating,

$$
A=
\left[
2x^2-\frac{2x^3}{3}
\right]_0^2

Evaluating at $x=2$,

$$
2(2)^2-\frac{2(2)^3}{3}
=
8-\frac{16}{3}

\frac{8}{3}
$$

Evaluating at $x=0$,

$$
0
$$

Therefore,

$$
\boxed{

}

$$
y=c^2-x^2

$$

=

2c^2x-\frac23x^3

At $x=-c$,

-\frac43c^3

\frac43c^3-
\left(-\frac43c^3\right)

$$
A=
\frac83c^3

576,
