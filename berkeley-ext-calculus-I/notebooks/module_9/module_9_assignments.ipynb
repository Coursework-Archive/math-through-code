```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle


def draw_fence(width, height, interior_fences=3):
    fig, ax = plt.subplots(figsize=(5, 3))

    # Outer rectangle
    ax.add_patch(
        Rectangle((0, 0), width, height,
                  fill=False, linewidth=2)
    )

    # Interior fences
    sections = interior_fences + 1
    spacing = width / sections

    for i in range(1, sections):
        x = i * spacing
        ax.plot([x, x], [0, height],
                color="black", linewidth=1.5)

    # Area
    area = width * height

    # Labels
    ax.text(width / 2, -height * 0.15,
            f"{width}", ha="center")

    ax.text(-width * 0.08, height / 2,
            f"{height}", va="center")

    ax.text(width * 1.08, height / 2,
            f"{area:,.1f} ft²",
            va="center")

    # Formatting
    ax.set_aspect("equal")
    ax.axis("off")
    plt.show()


# Estimated layouts
draw_fence(250, 50)
draw_fence(125, 100)
draw_fence(75, 120)
```

Use $a=3$ and $b=2$.

$$

$$
A(x)=4x\left(2\sqrt{1-\frac{x^2}{9}}\right)

$$

Differentiate using the product rule.

Let

$$
u=8x

and

$$

$$
u'=8

$$
v=\left(1-\frac{x^2}{9}\right)^{1/2}

Move one term:

$$

Multiply both sides by $\sqrt{1-\frac{x^2}{9}}$:
