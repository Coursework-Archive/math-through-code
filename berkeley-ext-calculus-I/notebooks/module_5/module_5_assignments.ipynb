## Question 76

Simplify:

$$
\tan(\sin^{-1} x)
$$

Let

$$
\theta = \sin^{-1}(x)
$$

Then

$$
\sin(\theta) = x
$$

Using a right triangle definition of sine

$$
\sin(\theta) = \frac{\text{opposite}}{\text{hypotenuse}}
$$

Choose

$$
\text{opposite} = x, \quad \text{hypotenuse} = 1
$$

Using the Pythagorean theorem

$$
1^2 = x^2 + (\text{adjacent})^2
$$

so

$$
\text{adjacent} = \sqrt{1 - x^2}
$$

Now compute tangent

$$
\tan(\theta) = \frac{\text{opposite}}{\text{adjacent}}
$$

Thus

$$
\tan(\theta) = \frac{x}{\sqrt{1 - x^2}}
$$

Substitute $\theta = \sin^{-1}(x)$

$$
\boxed{\frac{x}{\sqrt{1 - x^2}}}
$$

## Question 40

Given the curve
$$
x^2 + xy + y^2 = 3,
$$
find $y'$ and $y''$

Differentiate implicitly:
$$
\frac{d}{dx}(x^2) + \frac{d}{dx}(xy) + \frac{d}{dx}(y^2) = \frac{d}{dx}(3)
$$
$$
2x + (x y' + y) + 2y y' = 0
$$
$$
2x + y + x y' + 2y y' = 0
$$
$$
(x+2y)y' = -(2x+y)
$$
$$
y' = -\frac{2x+y}{x+2y}
$$

Now differentiate
$$
y' = -\frac{2x+y}{x+2y}
$$
using the quotient rule. Let
$$
u=2x+y, \qquad v=x+2y
$$
Then
$$
u'=2+y', \qquad v'=1+2y'
$$
So
$$
y'' = -\frac{(x+2y)(2+y')-(2x+y)(1+2y')}{(x+2y)^2}
$$

Substitute
$$
y'=-\frac{2x+y}{x+2y}
$$
into the numerator:
$$
2+y' = 2-\frac{2x+y}{x+2y}
= \frac{2(x+2y)-(2x+y)}{x+2y}
= \frac{3y}{x+2y}
$$
and
$$
1+2y' = 1-2\frac{2x+y}{x+2y}
= \frac{x+2y-4x-2y}{x+2y}
= -\frac{3x}{x+2y}
$$

Thus
$$
y''=
-\frac{(x+2y)\left(\frac{3y}{x+2y}\right)-(2x+y)\left(-\frac{3x}{x+2y}\right)}{(x+2y)^2}
$$
$$
=
-\frac{3y+\frac{3x(2x+y)}{x+2y}}{(x+2y)^2}
$$

Combine terms:
$$
y''=
-\frac{3y(x+2y)+3x(2x+y)}{(x+2y)^3}
$$
$$
=
-\frac{3xy+6y^2+6x^2+3xy}{(x+2y)^3}
$$
$$
=

## Question 51

Show that the tangent line to the circle
$$
x^2+y^2=r^2
$$
at any point $P=(x,y)$ is perpendicular to the radius drawn to $P$

Differentiate implicitly:
$$
2x+2y\,y'=0
$$
$$
2y\,y'=-2x
$$
$$
y'=-\frac{x}{y}
$$

So the slope of the tangent line at $P=(x,y)$ is
$$
m_{\text{tan}}=-\frac{x}{y}
$$

The radius from the origin $(0,0)$ to the point $P=(x,y)$ has slope
$$
m_{\text{rad}}=\frac{y-0}{x-0}=\frac{y}{x}
$$

Now multiply the slopes:
$$
m_{\text{tan}}\,m_{\text{rad}}
=
\left(-\frac{x}{y}\right)\left(\frac{y}{x}\right)
=-1.
$$

Since the slopes are negative reciprocals, the lines are perpendicular

$$
\boxed{\,m_{\text{tan}}=-\frac{x}{y}\,}
$$

$$
\boxed{\,m_{\text{rad}}=\frac{y}{x}\,}
$$

$$
\boxed{\,m_{\text{tan}}m_{\text{rad}}=-1\,}
$$

$$
\boxed{\,\text{The tangent line is perpendicular to the radius.}\,}
$$
