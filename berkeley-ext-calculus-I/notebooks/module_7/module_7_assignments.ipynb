## Question 26

Given:
$$
y = e^x, \quad x = 0, \quad \Delta x = 0.5
$$

Differential:
$$
dy = e^x dx
$$

$$
dy = 1 \cdot 0.5 = 0.5
$$

Actual change:
$$
\Delta y = e^{0.5} - 1 \approx 1.64872 - 1 = 0.64872
$$

$$
\boxed{dy = 0.5,\quad \Delta y \approx 0.64872}
$$



```python
import numpy as np
import matplotlib.pyplot as plt

# Section 3.10 - Question 26
# y = e^x, x = 0, dx = Δx = 0.5

x0 = 0
dx = 0.5

f = lambda x: np.exp(x)
fp = lambda x: np.exp(x)

y0 = f(x0)
x1 = x0 + dx
y1 = f(x1)

dy = fp(x0) * dx
delta_y = y1 - y0

# Tangent line at x = 0
L = lambda x: y0 + fp(x0) * (x - x0)

# Full graph range so lines reach the graph edges
x_vals = np.linspace(-0.3, 0.8, 400)
y_vals = f(x_vals)
tangent_vals = L(x_vals)

plt.figure(figsize=(8, 6))

# Function and tangent line
plt.plot(x_vals, y_vals, label=r"$f(x)=e^x$", linewidth=2)
plt.plot(
    x_vals,
    tangent_vals,
    "--",
    label=r"Tangent line $L(x)=1+x$",
    linewidth=2,
)

# Points
plt.scatter([x0, x1], [y0, y1], zorder=5)
plt.scatter([x1], [L(x1)], zorder=5)

# dx horizontal segment
plt.plot([x0, x1], [y0, y0], linewidth=3)
plt.text((x0 + x1) / 2, y0 - 0.08, r"$dx=\Delta x$", ha="center")

# dy vertical segment: from f(x0) to tangent line value
plt.plot([x1, x1], [y0, L(x1)], linewidth=3)
plt.text(x1 - 0.08, (y0 + L(x1)) / 2, r"$dy=0.5$", va="center", ha="right")

# Δy vertical segment: from f(x0) to actual curve value
plt.plot([x1 + 0.04, x1 + 0.04], [y0, y1], linewidth=3)
plt.text(
    x1 + 0.07,
    (y0 + y1) / 2,
    r"$\Delta y \approx 0.64872$",
    va="center",
    ha="left",
)

# Point label moved away from overlap
plt.annotate(
    r"$(0,1)$",
    xy=(x0, y0),
    xytext=(0.08, 0.88),
    arrowprops=dict(arrowstyle="->"),
)

plt.annotate(
    r"$(0.5,e^{0.5})$",
    xy=(x1, y1),
    xytext=(0.38, 1.78),
    arrowprops=dict(arrowstyle="->"),
)


plt.title(r"Question 26: $dy$ vs $\Delta y$ for $f(x)=e^x$")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.xlim(-0.3, 0.8)
plt.ylim(0.65, 2.1)

plt.show()
```
