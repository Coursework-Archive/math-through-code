## Question 46

Given

$$
f(x)=36x+3x^2-2x^3
$$

First derivative:

$$
f'(x)=36+6x-6x^2
$$

Factor:

$$
f'(x)=-6(x^2-x-6)
$$

$$
=-6(x-3)(x+2)
$$

Critical numbers:

$$
x=-2,\quad x=3
$$

Test intervals:

- $(-\infty,-2)$ gives $f'(x)<0$
- $(-2,3)$ gives $f'(x)>0$
- $(3,\infty)$ gives $f'(x)<0$

Therefore,

$$
\boxed{
\text{Increasing on } (-2,3)
}
$$

$$
\boxed{
\text{Decreasing on } (-\infty,-2)\cup(3,\infty)
}
$$

At $x=-2$:

$$
-\to+
$$

so there is a local minimum.

At $x=3$:

$$
+\to-
$$

so there is a local maximum.

Find function values:

$$
f(-2)=36(-2)+3(-2)^2-2(-2)^3
$$

$$
=-72+12+16
$$

$$
=-44
$$

Thus local minimum point:

$$
\boxed{(-2,-44)}
$$

Now evaluate:

$$
f(3)=36(3)+3(3)^2-2(3)^3
$$

$$
=108+27-54
$$

$$
=81
$$

Thus local maximum point:

$$
\boxed{(3,81)}
$$

---

Second derivative:

$$
f''(x)=6-12x
$$

Set equal to zero:

$$
6-12x=0
$$

$$
x=\frac12
$$

Test concavity:

- $(-\infty,\frac12)$ gives $f''(x)>0$
- $\left(\frac12,\infty\right)$ gives $f''(x)<0$

Therefore,

$$
\boxed{
\text{Concave up on } \left(-\infty,\frac12\right)
}
$$

$$
\boxed{
\text{Concave down on } \left(\frac12,\infty\right)
}
$$

Find inflection point:

$$
f\left(\frac12\right)
=36\left(\frac12\right)
+3\left(\frac12\right)^2
-2\left(\frac12\right)^3

=18+\frac34-\frac14

$$
=\frac{37}{2}

Thus,

$$
\boxed{
\text{Inflection point } \left(\frac12,\frac{37}{2}\right)

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-6, 7, 1000)
y = 36*x + 3*x**2 - 2*x**3

plt.figure(figsize=(7,5))

plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

plt.plot(x, y)

# Critical points
plt.scatter([-2, 3], [-44, 81])
plt.text(-2, -44, "(-2,-44)")
plt.text(3, 81, "(3,81)")

# Inflection point
plt.scatter([0.5], [37/2])
plt.text(0.5, 37/2, "(1/2,37/2)")

plt.grid(True)
plt.show()
```

## Question 42

Given

$$
y=(1-x)e^x
$$

**(E)** Intervals of increase or decrease:

Use the product rule.

Let

$$
f=1-x,\qquad g=e^x
$$

Then

$$
f'=-1,\qquad g'=e^x
$$

So,

$$
y'=(-1)e^x+(1-x)e^x
$$

$$
y'=-e^x+e^x-xe^x
$$

$$
y'=-xe^x
$$

Since $e^x>0$ for all $x$, the sign depends on $-x$.

Critical number:

$$
-xe^x=0
$$

$$
x=0
$$

Sign chart:

$$
(-\infty,0):+
$$

$$
(0,\infty):-
$$

Therefore,

$$
\boxed{\text{Increasing on }(-\infty,0)}
$$

$$
\boxed{\text{Decreasing on }(0,\infty)}
$$

---

**(F)** Local maximum and minimum:

At $x=0$, $y'$ changes from positive to negative, so there is a local maximum.

$$
y(0)=(1-0)e^0
$$

$$
y(0)=1
$$

Therefore,

$$
\boxed{\text{Local maximum at }(0,1)}
$$

$$
\boxed{\text{No local minimum}}
$$

---

**(G)** Concavity and points of inflection:

Start with

$$
y'=-xe^x
$$

Use the product rule again.

$$
y''=-xe^x-e^x
$$

$$
y''=-e^x(x+1)
$$

Set $y''=0$:

$$
-e^x(x+1)=0
$$

Since

$$
e^x\neq 0
$$

we solve:

$$
x+1=0
$$

$$
x=-1
$$

Sign chart:

$$
(-\infty,-1):+
$$

$$
(-1,\infty):-
$$

Therefore,

$$
\boxed{\text{Concave up on }(-\infty,-1)}
$$

$$
\boxed{\text{Concave down on }(-1,\infty)}
$$

Find the point of inflection:

$$
y(-1)=(1-(-1))e^{-1}
$$

$$
y(-1)=2e^{-1}
$$

$$
y(-1)=\frac{2}{e}
$$

Therefore,

$$
\boxed{\text{Point of inflection at }\left(-1,\frac{2}{e}\right)}
$$
